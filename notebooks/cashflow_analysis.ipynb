# Cash Flow Analytics & Capital Allocation Analysis
**Nifty 100 Financial Intelligence Platform — Sprint 2 Day 11**

This notebook provides visual insights into:
1. **Capital Allocation Pattern Distribution** across Nifty 100 companies.
2. **Free Cash Flow (FCF)** distribution and outlier breakdown.
3. **CFO Quality Score** distribution (Operating Cash Flow / Net Profit).
4. **CapEx Intensity** classification breakdown.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load generated output CSVs
base_dir = Path("..")
df_summary = pd.read_csv(base_dir / "output" / "cashflow_summary.csv")
df_stats = pd.read_csv(base_dir / "output" / "capital_pattern_statistics.csv")

df_summary.head()
```

## 1. Capital Allocation Pattern Distribution

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df_stats, x="pattern_label", y="count", palette="viridis", legend=False)
plt.title("Capital Allocation Pattern Frequency Distribution (Nifty 100)", fontsize=14, fontweight="bold")
plt.xlabel("Capital Allocation Pattern", fontsize=12)
plt.ylabel("Count of Company-Years", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 2. Free Cash Flow (FCF) Distribution

```python
plt.figure(figsize=(10, 5))
fcf_clean = df_summary["free_cash_flow"].dropna()
q_low, q_high = fcf_clean.quantile(0.05), fcf_clean.quantile(0.95)
fcf_trimmed = fcf_clean[(fcf_clean >= q_low) & (fcf_clean <= q_high)]

sns.histplot(fcf_trimmed, kde=True, color="#2ca02c", bins=35)
plt.axvline(fcf_trimmed.median(), color="red", linestyle="--", label=f"Median FCF: ₹{fcf_trimmed.median():.0f} Cr")
plt.title("Free Cash Flow Distribution (₹ Cr) — 90% Central Range", fontsize=14, fontweight="bold")
plt.xlabel("Free Cash Flow (₹ Cr)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. CFO Quality Distribution

```python
plt.figure(figsize=(10, 5))
cfo_qual = df_summary["cfo_quality_period"].dropna()
cfo_trimmed = cfo_qual[(cfo_qual >= 0) & (cfo_qual <= 3.0)]

sns.histplot(cfo_trimmed, kde=True, color="#1f77b4", bins=30)
plt.axvline(1.0, color="green", linestyle="--", label="High Quality Threshold (1.0)")
plt.axvline(0.5, color="orange", linestyle="--", label="Moderate / Accrual Risk Threshold (0.5)")
plt.title("CFO Quality Ratio (OCF / Net Profit) Distribution", fontsize=14, fontweight="bold")
plt.xlabel("CFO Quality Ratio", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```
